```python
def persian_to_english_numerals(text):
    if pd.isna(text) or not isinstance(text, str):
        return text
    persian_to_english_map = {
        '۰': '0', '۱': '1', '۲': '2', '۳': '3', '۴': '4',
        '۵': '5', '۶': '6', '۷': '7', '۸': '8', '۹': '9'
    }
    translated_text = ''.join([persian_to_english_map.get(char, char) for char in text])
    return translated_text
```

```python
def clean_and_convert_numeric_column(df, column_name):
    if column_name in df.columns:
        # Ensure the column is string type for processing
        df[column_name] = df[column_name].astype(str)
        # Apply Persian to English numeral conversion using the existing function
        df[column_name] = df[column_name].apply(persian_to_english_numerals)
        # Remove non-numeric characters (e.g., 'تومان', 'متر', commas, spaces) but keep dots for decimals
        df[column_name] = df[column_name].str.replace(r'[^۰-۹0-9.]', '', regex=True)
        # Convert to numeric, coercing errors to NaN
        df[column_name] = pd.to_numeric(df[column_name], errors='coerce')
    return df

# List of columns to clean and convert
columns_to_process = [
    'monthly_rent',
    'floor',
    'total_price',
    'price_per_meter',
    'area',
    'land_area',
    'deposit',
    'rooms',
    'builded_date'
]

# Apply the cleaning and conversion to each specified column
for col in columns_to_process:
    df_cleaned = clean_and_convert_numeric_column(df_cleaned, col)

print("DataFrame after converting Persian numerals and cleaning numeric columns:")
display(df_cleaned.head())
```

### Extract location

```python
import re

def extract_location_levels(location_text):
    if pd.isna(location_text) or not isinstance(location_text, str):
        return None, None

    # Find the part after 'در ' (in) which typically precedes the location info
    match = re.search(r'در\s*(.*)', location_text)
    if match:
        location_parts_str = match.group(1).strip()
        # Split by common delimiters like comma or '،'
        location_levels = [part.strip() for part in re.split(r'[،,]', location_parts_str) if part.strip()]

        level1 = location_levels[0] if len(location_levels) > 0 else None
        level2 = location_levels[1] if len(location_levels) > 1 else None
        return level1, level2
    return None, None

# Apply the function to the 'location' column
df_cleaned[['city', 'district']] = df_cleaned['location'].apply(lambda x: pd.Series(extract_location_levels(x)))

# Convert 'city' and 'district' to categorical type for LightGBM
df_cleaned['city'] = df_cleaned['city'].astype('category')
df_cleaned['district'] = df_cleaned['district'].astype('category')

print("Extracted location levels:")
display(df_cleaned[['location', 'city', 'district']].head())
```

### Remove outliers data IQR

```python
import numpy as np
import pandas as pd

# 1. Initial count for percentage calculation
initial_count = len(df_cleaned)

# Define bounds
price_min = 1.0e+08
price_max = 1.0e+13
area_max = 500
age_max = 100
floor_max_filter = 15.0 # Changed to filter threshold

# Build the initial filtering condition components
filter_conditions = [
    (df_cleaned['total_price'] >= price_min),
    (df_cleaned['total_price'] <= price_max),
    (df_cleaned['age'] <= age_max),
    (df_cleaned['area'] <= area_max)
]

# Add floor filter conditionally, if 'floor' column exists
if 'floor' in df_cleaned.columns:
    # Now filtering for floor values up to floor_max_filter
    filter_conditions.append(df_cleaned['floor'] <= floor_max_filter)
    filter_conditions.append(df_cleaned['floor'] >= -1) # Keep -1 for missing or basement floors
else:
    print("Warning: 'floor' column not found in df_cleaned. Skipping initial floor filter.")

# Combine all filter conditions
initial_filter = pd.Series(True, index=df_cleaned.index)
for cond in filter_conditions:
    initial_filter = initial_filter & cond

df_filtered = df_cleaned[initial_filter].copy()

# 2. Log Transformation
df_filtered['log_total_price'] = np.log1p(df_filtered['total_price'])

# 3. IQR-based Outlier Removal for Log Price
Q1_price = df_filtered['log_total_price'].quantile(0.25)
Q3_price = df_filtered['log_total_price'].quantile(0.75)
IQR_price = Q3_price - Q1_price
lower_bound_price = Q1_price - 1.5 * IQR_price
upper_bound_price = Q3_price + 1.5 * IQR_price

df_filtered = df_filtered[(df_filtered['log_total_price'] >= lower_bound_price) &
                         (df_filtered['log_total_price'] <= upper_bound_price)].copy()

# 4. No separate clipping for Floor as it's now filtered earlier

# 5. Calculate Filtering Statistics
# These statistics will now reflect rows removed by initial hard bounds and log_total_price IQR,
# but not by floor outliers, as they are clipped instead of removed.
final_count = len(df_filtered)
removed_count = initial_count - final_count
removed_percentage = (removed_count / initial_count) * 100

print(f"Initial data points: {initial_count}")
print(f"Data points after filtering (price, age, area, and floor): {final_count}")
print(f"Total rows removed by filtering: {removed_count}")
print(f"Percentage of rows removed by filtering: {removed_percentage:.2f}%")

df_filtered = df_filtered.drop(columns=['total_price'], errors='ignore')

# 6. Correlation Prep
df_corr_prep = df_filtered.copy()
material_categorical_cols = ['cabinet_type', 'floor_type', 'wall_type', 'cooler_type']
df_corr_prep = pd.get_dummies(df_corr_prep, columns=[c for c in material_categorical_cols if c in df_corr_prep.columns], dtype=int)

# 7. Calculate correlations
features_to_corr = ['rooms', 'floor', 'area', 'age', 'parking', 'elevator', 'storage',
                    'has_yard', 'has_balcony', 'has_cctv', 'has_kanaf_ceiling',
                    'has_double_glazed_windows', 'near_mosque_school', 'property_age_status',
                    'is_raw'] # Added new features

ohe_cols = [col for col in df_corr_prep.columns if any(col.startswith(f'{prefix}_') for prefix in material_categorical_cols)]
all_features_for_corr = [f for f in features_to_corr if f in df_corr_prep.columns] + ohe_cols + ['log_total_price']

new_corr = df_corr_prep[all_features_for_corr].corr()['log_total_price'].sort_values(ascending=False)
print("\nCorrelations with log_total_price:")
print(new_corr)
```

```python
import torch
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from category_encoders import TargetEncoder
from sklearn.feature_extraction.text import TfidfVectorizer # Import TfidfVectorizer

# Create combined feature: city and district
df_filtered['city_district'] = df_filtered['city'].astype(str) + '_' + df_filtered['district'].astype(str)

# Define features
# Removed 'city' and 'district' as standalone features, added 'city_district'
categorical_features_cols = ['cabinet_type', 'floor_type', 'wall_type', 'cooler_type', 'property_age_status', 'city_district']
binary_features = [
    'has_yard', 'has_balcony', 'has_cctv', 'has_kanaf_ceiling',
    'has_double_glazed_windows', 'near_mosque_school',
    'parking', 'elevator', 'storage',
    'is_raw'
]
numeric_features = ['rooms', 'floor', 'area', 'age']

# Reconstruct the 'features' list to include the new combined feature
features = numeric_features + categorical_features_cols + binary_features
target = 'log_total_price'

X = df_filtered[features].copy()
y = df_filtered[target].copy()

# --- Start: Add TF-IDF features for title ---
# Ensure tfidf_vectorizer_title is available from previous cells
# Re-apply TF-IDF on the filtered DataFrame's title
tfidf_matrix_title_filtered = tfidf_vectorizer_title.transform(df_filtered['title'].fillna(''))

# Create a DataFrame from the sparse TF-IDF matrix
# Align index with df_filtered to ensure correct concatenation
tfidf_df_title_filtered = pd.DataFrame(
    tfidf_matrix_title_filtered.toarray(),
    columns=[f'tfidf_title_{col}' for col in tfidf_vectorizer_title.get_feature_names_out()],
    index=df_filtered.index # Important: Align indices
)

# Concatenate TF-IDF features to X
X = pd.concat([X, tfidf_df_title_filtered], axis=1)
# --- End: Add TF-IDF features for title ---

# --- Start: Add TF-IDF features for description ---
# Initialize TfidfVectorizer for description
tfidf_vectorizer_description = TfidfVectorizer(
    max_features=300,
    min_df=5,
    max_df=0.7,
    ngram_range=(1, 2),
    sublinear_tf=True
)

# Fit and transform the 'description_processed' column
tfidf_matrix_description = tfidf_vectorizer_description.fit_transform(df_filtered['description_processed'].fillna(''))

# Create a DataFrame from the sparse TF-IDF matrix
tfidf_df_description = pd.DataFrame(
    tfidf_matrix_description.toarray(),
    columns=[f'tfidf_description_{col}' for col in tfidf_vectorizer_description.get_feature_names_out()],
    index=df_filtered.index
)

# Concatenate TF-IDF features to X
X = pd.concat([X, tfidf_df_description], axis=1)
# --- End: Add TF-IDF features for description ---


# Reduce cardinality for 'city_district' by keeping top N and labeling others as 'rare_location'
# This replaces the previous 'district' specific cardinality reduction
if 'city_district' in X.columns:
    top_100_city_districts = X['city_district'].value_counts().nlargest(160).index
    X['city_district'] = X['city_district'].apply(lambda x: x if x in top_100_city_districts else 'rare_location')
    # Ensure 'rare_location' is also a category
    X['city_district'] = X['city_district'].astype('category')

# Convert specified categorical columns to 'category' dtype
for col in categorical_features_cols:
    X[col] = X[col].astype('category')

# Replace -1 with NaN for imputation (applied before target encoding for numerical/binary features)
cols_to_impute = numeric_features + binary_features
for col in cols_to_impute:
    X[col] = X[col].replace(-1, np.nan)

# Train-Validation-Test Split
X_train_val, X_test, y_train_val, y_test = train_test_split(X, y.values, test_size=0.15, random_state=42) # 15% for test
X_train, X_val, y_train, y_val = train_test_split(X_train_val, y_train_val, test_size=0.20, random_state=42) # 20% of remaining for validation (approx 17% of total)

# Target Encoding for 'city_district'
# Removed 'city' and 'district', added 'city_district'
cols_to_target_encode = ['city_district']
te = TargetEncoder(cols=cols_to_target_encode)

# Fit and transform X_train
X_train[cols_to_target_encode] = te.fit_transform(X_train[cols_to_target_encode], y_train)
# Transform X_val and X_test
X_val[cols_to_target_encode] = te.transform(X_val[cols_to_target_encode])
X_test[cols_to_target_encode] = te.transform(X_test[cols_to_target_encode])

# Imputation for remaining NaNs (e.g., in numeric features or potential NaNs from TargetEncoder)
# This will handle NaNs in X_train, X_val, and X_test.
train_medians = X_train.median(numeric_only=True)
X_train = X_train.fillna(train_medians)
X_val = X_val.fillna(train_medians)
X_test = X_test.fillna(train_medians)

print("Data prepared.")
print("X_train shape:", X_train.shape)
print("X_val shape:", X_val.shape)
print("X_test shape:", X_test.shape)

print(f"Number of unique 'city_district' categories in full dataset (before split): {X['city_district'].nunique()}")
```

## Mean Absolute Percentage Error (MAPE)

```python
import numpy as np
import lightgbm as lgb

# 1. Train the model to ensure y_pred is available
lgbm_model = lgb.LGBMRegressor(objective='regression_l1',
                               n_estimators=1000,
                               learning_rate=0.01,
                               num_leaves=40,
                               random_state=42,
                               n_jobs=-1)

lgbm_model.fit(X_train, y_train.ravel(),
               eval_set=[(X_val, y_val.ravel())],
               callbacks=[lgb.early_stopping(100, verbose=False)])

# Generate predictions for the test set
y_pred = lgbm_model.predict(X_test)

# 2. Convert log_total_price back to original scale
actual_original_prices_test = np.expm1(y_test.flatten())
predicted_original_prices_test = np.expm1(y_pred.flatten())

# 3. Calculate MAPE
epsilon = 1e-10
absolute_percentage_errors = np.abs((actual_original_prices_test - predicted_original_prices_test) / (actual_original_prices_test + epsilon)) * 100
mean_absolute_percentage_error = np.mean(absolute_percentage_errors)

print(f"Mean Absolute Percentage Error (MAPE) on Test Set: {mean_absolute_percentage_error:.2f}%")
```
